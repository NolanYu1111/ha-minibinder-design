# Docking Output Parser & Binder Evaluation

This notebook processes docking scores (Rosetta/Vina output logs) and compares designed candidates to wild-type VH5-9-1 baseline scores.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set base paths
BASE_DIR = os.path.dirname(os.getcwd())
print("Project Base Directory:", BASE_DIR)
```

## 1. Load Rosetta or Vina Docking Energy Results

```python
# Placeholder: load or generate mock docking results table
# Scores include: design_id, target_strain, docking_score_kcal_mol, sasa_buried_percent

import numpy as np
np.random.seed(10)

strains = ['4WE4', '4WE5', '4ZCJ', '8TJ7', '2VIU', '2HMG']
designs = ['WT_VH5-9-1', 'Sandwich-1', 'Sandwich-2', 'Sandwich-3', 'HydroStack-7B', 'AroClamp-2F']

rows = []
for design in designs:
    for strain in strains:
        base_score = -9.5 if design == 'WT_VH5-9-1' else -12.4
        score = np.random.normal(base_score, 1.2)
        rows.append({
            'design_id': design,
            'target_strain': strain,
            'docking_score_kcal_mol': round(score, 2),
            'sasa_buried_percent': round(np.random.normal(15 if design == 'WT_VH5-9-1' else 25, 3), 1)
        })

df_docking = pd.DataFrame(rows)
df_docking.head()
```

```python
# Visualize docking energy comparison
plt.figure(figsize=(10, 6))
sns.barplot(data=df_docking, x='design_id', y='docking_score_kcal_mol', hue='target_strain')
plt.title('Predicted Docking Energy (kcal/mol) Across Diverse HA Strains')
plt.ylabel('Docking score (kcal/mol)')
plt.xlabel('Antibody Variant')
plt.legend(title='HA Strain')
plt.tight_layout()
plt.show()
```
