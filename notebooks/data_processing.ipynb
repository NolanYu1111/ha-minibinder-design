# Data Processing & Curation Analysis

This notebook analyzes the curated HA structures, runs epitope solvent accessible surface area (SASA) checks, and visualizes glycan shielding distances.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set project paths
BASE_DIR = os.path.dirname(os.getcwd())
METADATA_FILE = os.path.join(BASE_DIR, 'data', 'processed', 'ha_structures_curated.csv')
SASA_FILE = os.path.join(BASE_DIR, 'data', 'processed', 'epitope_sasa_summary.csv')

print("Project Base Directory:", BASE_DIR)
```

```python
# Load data
df_meta = pd.read_csv(METADATA_FILE)
df_sasa = pd.read_csv(SASA_FILE)
display(df_sasa.head())
```

```python
# Plot distribution of Total Motif SASA across panels
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_sasa, x='Panel', y='Total_Motif_SASA')
plt.title('Epitope SASA (Pro+Trp) Distribution Across Panels')
plt.ylabel('SASA (A^2)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
