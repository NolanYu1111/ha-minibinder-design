# Integrated Scoring Metrics & Figure Generation

This notebook integrates predicted binder classification scores, docking scores, structural confidence, and FoldX stability values to create the final paper-ready figures.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set base paths
BASE_DIR = os.path.dirname(os.getcwd())
print("Project Base Directory:", BASE_DIR)
```

## 1. Load Combined Metrics Table

```python
# Generate mock data for composite scores
np.random.seed(50)
designs = [f'Sandwich-{i:02d}' for i in range(1, 21)] + ['WT_VH5-9-1']

rows = []
for design in designs:
    is_wt = design == 'WT_VH5-9-1'
    rows.append({
        'design_id': design,
        'docking_score': np.random.normal(-9.5 if is_wt else -13.2, 0.8),
        'sasa_buried': np.random.normal(520 if is_wt else 710, 45),
        'foldx_ddG': np.random.normal(0.0 if is_wt else -1.8, 0.5),
        'pLDDT': np.random.normal(83 if is_wt else 91, 2.5),
        'aromatic_sandwich_score': np.random.uniform(0.3 if is_wt else 0.7, 0.95)
    })

df_metrics = pd.DataFrame(rows)
df_metrics.head()
```

```python
# Plot Aromatic Sandwich Score vs. Docking Energy
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_metrics, x='aromatic_sandwich_score', y='docking_score', size='sasa_buried', hue='pLDDT', palette='viridis', sizes=(50, 200))
plt.axvline(x=0.7, color='red', linestyle='--', label='Sandwich threshold')
plt.title('Aromatic Stacking Geometry vs. Docking Score')
plt.xlabel('Aromatic Sandwich Score')
plt.ylabel('Docking Score (kcal/mol)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
